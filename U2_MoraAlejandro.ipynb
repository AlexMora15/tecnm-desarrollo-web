```python
def validar_datos(prestamos):
    # Caso base: lista vacía es válida
    if len(prestamos) == 0:
        return True
    primero = prestamos[0]
    if not isinstance(primero, int) or isinstance(primero, bool) or primero < 0:
        raise ValueError(f"Dato inválido '{primero}'. Solo se permiten enteros no negativos.")
    # Paso recursivo
    return validar_datos(prestamos[1:])


def total_prestamos(prestamos):
    validar_datos(prestamos)
    # Caso base
    if len(prestamos) == 0:
        return 0
    # Paso recursivo
    return prestamos[0] + total_prestamos(prestamos[1:])


def semanas_activas(prestamos):
    validar_datos(prestamos)
    # Caso base
    if len(prestamos) == 0:
        return 0
    # Paso recursivo
    activa = 1 if prestamos[0] > 0 else 0
    return activa + semanas_activas(prestamos[1:])


def total_prestamos_iterativo(prestamos):
    total = 0
    for cantidad in prestamos:
        if not isinstance(cantidad, int) or isinstance(cantidad, bool) or cantidad < 0:
            raise ValueError(f"Dato inválido '{cantidad}'. Solo se permiten enteros no negativos.")
        total += cantidad
    return total
```

```python
pruebas = [
    ("Lista vacía", []),
    ("Una semana", [7]),
    ("Varias semanas", [3, 0, 5, 2, 0, 4]),
    ("Datos incorrectos", [2, -1, 3])
]

for nombre, lista in pruebas:
    print(f"--- {nombre}: {lista} ---")
    try:
        print(f"  Total préstamos (recursivo): {total_prestamos(lista)}")
        print(f"  Total préstamos (iterativo): {total_prestamos_iterativo(lista)}")
        print(f"  Semanas activas (recursivo): {semanas_activas(lista)}\n")
    except ValueError as err:
        print(f"  Error controlado: {err}\n")

        ## Conclusión

Desarrollar el análisis de préstamos mediante recursividad permitió comprender cómo un problema sobre listas puede dividirse en subproblemas cada vez más pequeños hasta alcanzar un caso base sencillo. La principal ventaja de la recursividad es la claridad y limpieza del código: al evitar ciclos y variables acumuladoras, la lógica matemática de sumar el primer elemento con el resto de la lista resulta directa, expresiva y fácil de verificar.

No obstante, al comparar este método con su equivalente iterativo, se observan desventajas importantes en el uso de recursos. Mientras que el enfoque iterativo trabaja con memoria constante $O(1)$, la recursividad apila cada llamada pendiente en memoria ($O(n)$), lo que aumenta el consumo de recursos y puede provocar un error de desbordamiento de pila si se procesaran registros de miles de semanas. Además, el corte de listas (`prestamos[1:]`) crea copias adicionales en cada llamada. En conclusión, la recursividad es ideal para problemas naturalmente jerárquicos, pero para recorridos lineales simples la iteración sigue siendo más eficiente.
```
