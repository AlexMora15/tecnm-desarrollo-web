```python
# ==========================================
# DEFINICIÓN DE NODOS PROPIOS (SIN LISTAS DE PYTHON)
# ==========================================

class NodoAccion:
    """Nodo para la Pila del historial de operaciones."""
    def __init__(self, accion):
        self.accion = accion
        self.siguiente = None

class NodoUsuario:
    """Nodo para la Cola de atención de turnos."""
    def __init__(self, nombre_usuario):
        self.nombre_usuario = nombre_usuario
        self.siguiente = None

class NodoPrestamo:
    """Nodo para la Lista Enlazada de préstamos activos."""
    def __init__(self, id_prestamo, usuario, libro):
        self.id_prestamo = id_prestamo
        self.usuario = usuario
        self.libro = libro
        self.siguiente = None

print("Nodos definidos correctamente.")
```

```python
# ==========================================
# MÓDULO 1: HISTORIAL DE OPERACIONES (PILA)
# ==========================================
class PilaHistorial:
    def __init__(self):
        self.cima = None

    # 4. Detectar cuando la pila esté vacía
    def esta_vacia(self):
        return self.cima is None

    # 1. Registrar acciones realizadas (Push)
    def registrar_accion(self, accion):
        nuevo_nodo = NodoAccion(accion)
        nuevo_nodo.siguiente = self.cima
        self.cima = nuevo_nodo
        print(f"[Pila] Acción registrada: '{accion}'")

    # 2. Consultar la última acción (Peek)
    def consultar_ultima(self):
        if self.esta_vacia():
            print("[Pila] El historial está vacío. No hay acciones por consultar.")
            return None
        print(f"[Pila] Última acción en la cima: '{self.cima.accion}'")
        return self.cima.accion

    # 3. Deshacer la última acción registrada (Pop)
    def deshacer_accion(self):
        if self.esta_vacia():
            print("[Pila] No se puede deshacer: el historial está vacío.")
            return None
        accion_deshecha = self.cima.accion
        self.cima = self.cima.siguiente
        print(f"[Pila] Acción deshecha: '{accion_deshecha}'")
        return accion_deshecha

print("Módulo 1 (Pila) listo.")
```

```python
# ==========================================
# MÓDULO 2: ATENCIÓN DE USUARIOS (COLA)
# ==========================================
class ColaTurnos:
    def __init__(self):
        self.frente = None
        self.final = None

    # 4. Detectar cuando la cola esté vacía
    def esta_vacia(self):
        return self.frente is None

    # 1. Registrar usuarios por orden de llegada (Enqueue)
    def encolar_usuario(self, nombre_usuario):
        nuevo_nodo = NodoUsuario(nombre_usuario)
        if self.esta_vacia():
            self.frente = nuevo_nodo
            self.final = nuevo_nodo
        else:
            self.final.siguiente = nuevo_nodo
            self.final = nuevo_nodo
        print(f"[Cola] Usuario formado en turno: {nombre_usuario}")

    # 2. Consultar al siguiente usuario (Front/Peek)
    def consultar_siguiente(self):
        if self.esta_vacia():
            print("[Cola] No hay usuarios esperando turno.")
            return None
        print(f"[Cola] Siguiente usuario en turno: {self.frente.nombre_usuario}")
        return self.frente.nombre_usuario

    # 3. Atender y retirar al primer usuario registrado (Dequeue)
    def atender_usuario(self):
        if self.esta_vacia():
            print("[Cola] No se puede atender: la fila está vacía.")
            return None
        usuario_atendido = self.frente.nombre_usuario
        self.frente = self.frente.siguiente
        if self.frente is None:
            self.final = None
        print(f"[Cola] Usuario atendido y retirado: {usuario_atendido}")
        return usuario_atendido

print("Módulo 2 (Cola) listo.")
```

```python
# ==========================================
# MÓDULO 3: PRÉSTAMOS ACTIVOS (LISTA ENLAZADA)
# ==========================================
class ListaPrestamos:
    def __init__(self):
        self.cabeza = None

    def esta_vacia(self):
        return self.cabeza is None

    # 1. Registrar préstamos
    def registrar_prestamo(self, id_prestamo, usuario, libro):
        nuevo_nodo = NodoPrestamo(id_prestamo, usuario, libro)
        if self.esta_vacia():
            self.cabeza = nuevo_nodo
        else:
            actual = self.cabeza
            while actual.siguiente is not None:
                actual = actual.siguiente
            actual.siguiente = nuevo_nodo
        print(f"[Lista] Préstamo registrado -> ID: {id_prestamo} | Usuario: {usuario} | Libro: '{libro}'")

    # 2. Consultar un préstamo
    def consultar_prestamo(self, id_prestamo):
        actual = self.cabeza
        while actual is not None:
            if actual.id_prestamo == id_prestamo:
                print(f"[Lista] Encontrado -> ID: {actual.id_prestamo} | Usuario: {actual.usuario} | Libro: '{actual.libro}'")
                return actual
            actual = actual.siguiente
        print(f"[Lista] Búsqueda fallida: No existe ningún préstamo con ID '{id_prestamo}'.")
        return None

    # 3. Eliminar un préstamo
    def eliminar_prestamo(self, id_prestamo):
        if self.esta_vacia():
            print("[Lista] No se puede eliminar: no hay préstamos activos.")
            return False

        # Caso 1: El préstamo a eliminar es el primero (cabeza)
        if self.cabeza.id_prestamo == id_prestamo:
            self.cabeza = self.cabeza.siguiente
            print(f"[Lista] Préstamo con ID '{id_prestamo}' eliminado correctamente.")
            return True

        # Caso 2: El préstamo está en medio o al final
        anterior = self.cabeza
        actual = self.cabeza.siguiente
        while actual is not None:
            if actual.id_prestamo == id_prestamo:
                anterior.siguiente = actual.siguiente
                print(f"[Lista] Préstamo con ID '{id_prestamo}' eliminado correctamente.")
                return True
            anterior = actual
            actual = actual.siguiente

        print(f"[Lista] No se pudo eliminar: ID '{id_prestamo}' no encontrado.")
        return False

    # 4. Mostrar todos los préstamos registrados
    def mostrar_prestamos(self):
        if self.esta_vacia():
            print("[Lista] No hay préstamos activos registrados.")
            return
        print("\n--- LISTADO DE PRÉSTAMOS ACTIVOS ---")
        actual = self.cabeza
        while actual is not None:
            print(f"  • ID: {actual.id_prestamo} | Usuario: {actual.usuario} | Libro: '{actual.libro}'")
            actual = actual.siguiente
        print("------------------------------------\n")

print("Módulo 3 (Lista Enlazada) listo.")
```

```python
# ==========================================
# EJECUCIÓN DE PRUEBAS SOLICITADAS
# ==========================================
print("=== 1. COMPROBACIÓN DE ESTRUCTURAS VACÍAS ===")
historial = PilaHistorial()
turnos = ColaTurnos()
prestamos = ListaPrestamos()

historial.consultar_ultima()
historial.deshacer_accion()
turnos.consultar_siguiente()
turnos.atender_usuario()
prestamos.mostrar_prestamos()
prestamos.eliminar_prestamo("P-99")

print("\n=== 2. PRUEBAS DE PILA (3 ACCIONES, DESHACER 2) ===")
historial.registrar_accion("Alta de libro: Cien años de soledad")
historial.registrar_accion("Préstamo registrado: P-01")
historial.registrar_accion("Actualización de datos de usuario")
historial.consultar_ultima()
historial.deshacer_accion()
historial.deshacer_accion()
historial.consultar_ultima()

print("\n=== 3. PRUEBAS DE COLA (ENCOLAR 3 USUARIOS, ATENDER 2) ===")
turnos.encolar_usuario("Ana López")
turnos.encolar_usuario("Carlos Pérez")
turnos.encolar_usuario("María Gómez")
turnos.consultar_siguiente()
turnos.atender_usuario()
turnos.atender_usuario()
turnos.consultar_siguiente()

print("\n=== 4. PRUEBAS DE LISTA ENLAZADA (4 PRÉSTAMOS, BUSCAR Y ELIMINAR) ===")
prestamos.registrar_prestamo("P-01", "Ana López", "Estructuras de Datos en Python")
prestamos.registrar_prestamo("P-02", "Carlos Pérez", "El Principito")
prestamos.registrar_prestamo("P-03", "María Gómez", "Clean Code")
prestamos.registrar_prestamo("P-04", "Luis Torres", "Algoritmos Básicos")
prestamos.mostrar_prestamos()

# Buscar existente e inexistente
prestamos.consultar_prestamo("P-03")
prestamos.consultar_prestamo("P-99")

# Eliminar un préstamo y mostrar lista actualizada
prestamos.eliminar_prestamo("P-02")
prestamos.mostrar_prestamos()
```

# Evidencia 4: Cuadro comparativo y análisis de eficiencia

## 1. Cuadro comparativo de estructuras lineales

| Estructura | Características principales | Representación en memoria | Operaciones básicas | Aplicaciones comunes |
| :--- | :--- | :--- | :--- | :--- |
| **Pila** | Sigue el principio **LIFO** (*Last In, First Out*). Todas las inserciones y eliminaciones ocurren por un solo extremo llamado *cima*. | Arreglo estático o nodos enlazados donde cada nodo apunta al elemento inferior. | `push` (apilar), `pop` (desapilar), `peek` (ver cima), `isEmpty`. | Historial de acciones (Deshacer/Ctrl+Z), navegación hacia atrás, evaluación de expresiones matemáticas. |
| **Cola simple** | Sigue el principio **FIFO** (*First In, First Out*). Se inserta por el *final* y se extrae por el *frente*. | Arreglo lineal o nodos enlazados con dos referencias: `frente` y `final`. | `enqueue` (encolar), `dequeue` (desencolar), `front` (ver frente). | Turnos de atención en ventanilla, colas de impresión, gestión de solicitudes en servidores. |
| **Cola circular** | Variante FIFO donde la última posición se conecta con la primera formando un anillo lógico para reutilizar espacios libres. | Arreglo de tamaño fijo con índices circulares (`(i + 1) % N`) o lista enlazada circular. | `enqueue`, `dequeue`, verificación de cola llena y vacía. | Búferes de transmisión de audio/video, planificación de procesos por turnos (*Round Robin*). |
| **Bicola (Deque)** | Cola de doble extremo que permite insertar y eliminar elementos tanto por el frente como por el final. | Arreglo circular o lista doblemente enlazada con punteros a ambos extremos. | `insertFront`, `insertRear`, `deleteFront`, `deleteRear`. | Historiales de navegación combinados, algoritmos de ventana deslizante, gestión de tareas con prioridad. |
| **Lista simplemente enlazada** | Colección dinámica de nodos; cada nodo contiene un dato y un único apuntador al siguiente nodo. | Nodos dispersos en memoria conectados unidireccionalmente desde `cabeza` hasta `None`. | Insertar, buscar, eliminar por valor/ID y recorrer hacia adelante. | Gestión de registros dinámicos (préstamos activos), listas de reproducción simples. |
| **Lista doblemente enlazada** | Cada nodo almacena un dato y dos apuntadores: uno al nodo siguiente y otro al nodo anterior. | Nodos dispersos en memoria con enlaces bidireccionales (`anterior` y `siguiente`). | Insertar al inicio/final, eliminar nodo directo, recorrer en ambos sentidos. | Editores de texto, carruseles de imágenes, historial del navegador (Atrás y Adelante). |
| **Lista circular** | El apuntador `siguiente` del último nodo apunta de regreso al primer nodo (`cabeza`) en lugar de apuntar a `None`. | Cadena cerrada de nodos sin inicio ni fin nulo marcado. | Insertar, eliminar y recorrer cíclicamente sin desbordamiento. | Juegos por turnos multijugador, listas de reproducción en bucle continuo. |

## 2. Análisis de complejidad temporal (Notación Big-O)

* **Pila (`PilaHistorial`):**
  * **Registrar acción (`registrar_accion` / *Push*):** **$O(1)$ (Tiempo constante).** Solo se crea un nuevo nodo y se actualiza la referencia `cima`, sin importar cuántos elementos existan en la pila.
  * **Deshacer acción (`deshacer_accion` / *Pop*):** **$O(1)$ (Tiempo constante).** Basta con mover el puntero `cima` hacia `cima.siguiente`, sin necesidad de recorrer la estructura.

* **Cola (`ColaTurnos`):**
  * **Registrar usuario (`encolar_usuario` / *Enqueue*):** **$O(1)$ (Tiempo constante).** Gracias a que se mantiene un puntero directo al `final` de la cola, el nuevo nodo se enlaza de inmediato sin recorrer la fila.
  * **Atender usuario (`atender_usuario` / *Dequeue*):** **$O(1)$ (Tiempo constante).** Se extrae directamente el nodo ubicado en `frente` y se recorre el puntero al siguiente nodo en un solo paso.

* **Lista Simplemente Enlazada (`ListaPrestamos`):**
  * **Consultar un préstamo (`consultar_prestamo`):** **$O(n)$ (Tiempo lineal).** En el peor de los casos (si el préstamo está al final o no existe), se deben revisar los $n$ nodos uno por uno desde `cabeza`.
  * **Eliminar un préstamo (`eliminar_prestamo`):** **$O(n)$ (Tiempo lineal).** Aunque reconectar los apuntadores toma $O(1)$, primero se requiere recorrer hasta $n$ nodos para localizar el ID del préstamo a eliminar.
