## Ejercicio 1. Árbol Binario de Búsqueda (ABB)

### Explicación del funcionamiento:
1. **Clase `NodoLibro`:** Representa cada libro dentro del árbol. Guarda su identificador numérico (`id_libro`) y dos referencias: `izquierda` (para IDs menores) y `derecha` (para IDs mayores).
2. **Inserción (`insertar`):** Compara el nuevo ID con el nodo actual. Si es menor, baja por la rama izquierda; si es mayor, baja por la derecha hasta encontrar un espacio vacío (`None`).
3. **Búsqueda (`buscar`):** Inicia en la raíz y va descartando la mitad del árbol en cada paso comparando si el ID buscado es menor o mayor al nodo actual.
4. **Eliminación (`eliminar`):** Localiza el nodo y maneja tres casos: si no tiene hijos se borra directo; si tiene un hijo, este ocupa su lugar; y si tiene dos hijos, se reemplaza por el valor mínimo de su subárbol derecho (sucesor inorden).
5. **Recorrido Inorden (`inorden`):** Visita primero el subárbol izquierdo, luego la raíz y al final el subárbol derecho, lo que imprime todos los IDs de libros ordenados de menor a mayor.

```python
# ==========================================================
# EJERCICIO 1: ÁRBOL BINARIO DE BÚSQUEDA (LIBROS)
# ==========================================================

class NodoLibro:
    def __init__(self, id_libro):
        self.id_libro = id_libro
        self.izquierda = None
        self.derecha = None

class ArbolBinarioBusqueda:
    def __init__(self):
        self.raiz = None

    # 1. Registrar (insertar) un identificador de libro
    def insertar(self, id_libro):
        if self.raiz is None:
            self.raiz = NodoLibro(id_libro)
        else:
            self._insertar_recursivo(self.raiz, id_libro)

    def _insertar_recursivo(self, actual, id_libro):
        if id_libro < actual.id_libro:
            if actual.izquierda is None:
                actual.izquierda = NodoLibro(id_libro)
            else:
                self._insertar_recursivo(actual.izquierda, id_libro)
        elif id_libro > actual.id_libro:
            if actual.derecha is None:
                actual.derecha = NodoLibro(id_libro)
            else:
                self._insertar_recursivo(actual.derecha, id_libro)
        else:
            print(f"[Aviso] El libro con ID {id_libro} ya existe en el árbol.")

    # 2. Buscar un identificador existente
    def buscar(self, id_libro):
        return self._buscar_recursivo(self.raiz, id_libro)

    def _buscar_recursivo(self, actual, id_libro):
        if actual is None:
            return False
        if actual.id_libro == id_libro:
            return True
        elif id_libro < actual.id_libro:
            return self._buscar_recursivo(actual.izquierda, id_libro)
        else:
            return self._buscar_recursivo(actual.derecha, id_libro)

    # 3. Eliminar un identificador del árbol
    def eliminar(self, id_libro):
        self.raiz = self._eliminar_recursivo(self.raiz, id_libro)

    def _eliminar_recursivo(self, actual, id_libro):
        if actual is None:
            return actual

        # Buscamos el nodo a eliminar
        if id_libro < actual.id_libro:
            actual.izquierda = self._eliminar_recursivo(actual.izquierda, id_libro)
        elif id_libro > actual.id_libro:
            actual.derecha = self._eliminar_recursivo(actual.derecha, id_libro)
        else:
            # Caso 1 y 2: Nodo sin hijos o con un solo hijo
            if actual.izquierda is None:
                return actual.derecha
            elif actual.derecha is None:
                return actual.izquierda

            # Caso 3: Nodo con dos hijos -> buscar el menor del subárbol derecho
            sucesor = self._obtener_minimo(actual.derecha)
            actual.id_libro = sucesor.id_libro
            actual.derecha = self._eliminar_recursivo(actual.derecha, sucesor.id_libro)

        return actual

    def _obtener_minimo(self, nodo):
        actual = nodo
        while actual.izquierda is not None:
            actual = actual.izquierda
        return actual

    # 4. Mostrar registros mediante recorrido Inorden (Izquierda -> Raíz -> Derecha)
    def recorrido_inorden(self):
        resultado = []
        self._inorden_recursivo(self.raiz, resultado)
        return resultado

    def _inorden_recursivo(self, actual, resultado):
        if actual is not None:
            self._inorden_recursivo(actual.izquierda, resultado)
            resultado.append(actual.id_libro)
            self._inorden_recursivo(actual.derecha, resultado)


# ==========================================================
# PRUEBAS DEL EJERCICIO 1
# ==========================================================
arbol_libros = ArbolBinarioBusqueda()

# 1. Registrar al menos 6 identificadores numéricos de libros
ids_libros = [50, 30, 70, 20, 40, 60, 80]
for id_l in ids_libros:
    arbol_libros.insertar(id_l)

print("1. Libros registrados (Recorrido Inorden ordenado):", arbol_libros.recorrido_inorden())

# 2. Buscar un identificador existente (y uno no existente para comprobar)
id_buscar = 40
encontrado = arbol_libros.buscar(id_buscar)
print(f"2. ¿El libro con ID {id_buscar} existe en el árbol?: {'Sí, encontrado' if encontrado else 'No encontrado'}")

# 3. Eliminar un identificador del árbol (por ejemplo, el 30 que tiene dos hijos)
id_borrar = 30
arbol_libros.eliminar(id_borrar)
print(f"3. Se eliminó el libro con ID {id_borrar}.")

# 4. Mostrar nuevamente el recorrido inorden tras la eliminación
print("4. Libros activos después de eliminar (Recorrido Inorden):", arbol_libros.recorrido_inorden())
```

## Ejercicio 2. Grafo de Bibliotecas

### Explicación del funcionamiento:
1. **Representación (`lista_adyacencia`):** Se utiliza un diccionario de Python donde cada clave es un vértice (biblioteca) y su valor es una lista con las bibliotecas con las que tiene conexión directa.
2. **Agregar vértice (`agregar_biblioteca`):** Crea una entrada nueva en el diccionario con una lista vacía de conexiones.
3. **Registrar conexión (`agregar_conexion`):** Como la red de préstamos interbibliotecarios funciona en ambos sentidos (grafo no dirigido), al conectar la biblioteca A con la B se añade B a la lista de A y A a la lista de B.
4. **Consultar vecinos (`consultar_vecinos`):** Devuelve directamente la lista de bibliotecas conectadas a la biblioteca consultada.

```python
# ==========================================================
# EJERCICIO 2: GRAFO DE BIBLIOTECAS (LISTA DE ADYACENCIA)
# ==========================================================

class GrafoBibliotecas:
    def __init__(self):
        self.lista_adyacencia = {}

    # 1. Crear un vértice (biblioteca)
    def agregar_biblioteca(self, nombre):
        if nombre not in self.lista_adyacencia:
            self.lista_adyacencia[nombre] = []
            print(f"[Grafo] Biblioteca agregada: '{nombre}'")
        else:
            print(f"[Aviso] La biblioteca '{nombre}' ya existe.")

    # 2. Registrar una conexión (arista no dirigida entre dos bibliotecas)
    def agregar_conexion(self, biblio1, biblio2):
        if biblio1 in self.lista_adyacencia and biblio2 in self.lista_adyacencia:
            if biblio2 not in self.lista_adyacencia[biblio1]:
                self.lista_adyacencia[biblio1].append(biblio2)
                self.lista_adyacencia[biblio2].append(biblio1)
                print(f"[Grafo] Conexión registrada: {biblio1} <---> {biblio2}")
        else:
            print("[Error] Una o ambas bibliotecas no existen en la red.")

    # 3. Mostrar todas las conexiones usando lista de adyacencia
    def mostrar_grafo(self):
        print("\n--- LISTA DE ADYACENCIA DE LA RED DE BIBLIOTECAS ---")
        for biblioteca, vecinos in self.lista_adyacencia.items():
            conexiones = ", ".join(vecinos) if vecinos else "Sin conexiones"
            print(f"  • {biblioteca} -> [{conexiones}]")
        print("----------------------------------------------------\n")

    # 4. Consultar los vecinos de una biblioteca específica
    def consultar_vecinos(self, nombre):
        if nombre in self.lista_adyacencia:
            vecinos = self.lista_adyacencia[nombre]
            print(f"Vecinos directos de '{nombre}': {vecinos}")
            return vecinos
        else:
            print(f"La biblioteca '{nombre}' no está registrada.")
            return []


# ==========================================================
# PRUEBAS DEL EJERCICIO 2
# ==========================================================
red_bibliotecas = GrafoBibliotecas()

# 1. Crear cuatro vértices
red_bibliotecas.agregar_biblioteca("Biblioteca Puerto Morelos")
red_bibliotecas.agregar_biblioteca("Biblioteca Cancún Centro")
red_bibliotecas.agregar_biblioteca("Biblioteca Playa del Carmen")
red_bibliotecas.agregar_biblioteca("Biblioteca Leona Vicario")

print()
# 2. Registrar al menos cuatro conexiones
red_bibliotecas.agregar_conexion("Biblioteca Puerto Morelos", "Biblioteca Cancún Centro")
red_bibliotecas.agregar_conexion("Biblioteca Puerto Morelos", "Biblioteca Playa del Carmen")
red_bibliotecas.agregar_conexion("Biblioteca Puerto Morelos", "Biblioteca Leona Vicario")
red_bibliotecas.agregar_conexion("Biblioteca Cancún Centro", "Biblioteca Leona Vicario")

# 3. Mostrar las conexiones utilizando la lista de adyacencia
red_bibliotecas.mostrar_grafo()

# 4. Consultar los vecinos de una biblioteca
red_bibliotecas.consultar_vecinos("Biblioteca Puerto Morelos")
red_bibliotecas.consultar_vecinos("Biblioteca Playa del Carmen")
```

## Análisis Final

**1. ¿Qué diferencia existe entre un árbol y un grafo?**  
Un **árbol** es una estructura jerárquica con un nodo principal llamado *raíz*, donde los elementos van de arriba hacia abajo (padres e hijos) y **no existen ciclos** ni caminos cerrados (hay un único camino entre dos nodos). En cambio, un **grafo** es una red más libre donde cualquier nodo (vértice) puede conectarse con cualquier otro sin seguir una jerarquía, por lo que **sí puede tener ciclos**, múltiples caminos entre nodos e incluso nodos aislados.

**2. ¿Para qué utilizarías un árbol binario de búsqueda?**  
Lo utilizaría cuando necesito almacenar datos que se consultan, insertan y eliminan constantemente y además requiero mantenerlos ordenados automáticamente —por ejemplo, un catálogo de libros por código, un directorio de matrículas de alumnos o un inventario de productos—, ya que permite encontrar cualquier registro muy rápido sin tener que revisar toda la colección uno por uno.

**3. ¿Qué ventaja tiene representar las bibliotecas mediante un grafo?**  
Permite modelar de manera natural cómo están interconectadas las distintas sedes entre sí (sin importar si forman ciclos o rutas alternativas), facilitando saber qué bibliotecas tienen convenio o ruta directa de intercambio de libros, detectar sedes aisladas y calcular las rutas más cortas o rápidas para trasladar ejemplares de una biblioteca a otra.
